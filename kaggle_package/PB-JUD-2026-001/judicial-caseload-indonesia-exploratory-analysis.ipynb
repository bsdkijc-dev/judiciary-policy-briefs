## 1. Research Introduction
This notebook reproduces the empirical analysis for *From Court Statistics to Better Justice Policy: Understanding Caseload, Case Clearance, and Pending Cases in Indonesia's Judiciary (2015–2025)*.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# 2. Dataset Loading (works both on Kaggle /kaggle/input/ and locally)
base = Path('/kaggle/input/indonesia-judiciary-caseload-2015-2025')
if not base.exists():
    base = Path('.')
df = pd.read_csv(base / 'judicial_caseload.csv')
src = pd.read_csv(base / 'source_register.csv')
dd = pd.read_csv(base / 'data_dictionary.csv')
val = pd.read_csv(base / 'validation_status.csv')
print(f'Loaded {len(df)} records across {len(src)} official sources.')
```

## 3. Data Dictionary & 4. Data Quality Checks


```python
print(dd[['column_name', 'data_type', 'unit']].to_string(index=False))
print('\nValidation Checks:')
print(val[['check_id', 'status', 'details']].to_string(index=False))
```

## 5. Missing-Value Analysis & 6. Descriptive Statistics


```python
print('Missing counts (zero-imputation policy):')
print(df[['opening_pending', 'incoming_cases', 'total_caseload', 'resolved_cases', 'pending_cases']].isna().sum())
ma = df[df['court_id'] == 'MA-RI'].sort_values('report_year').copy()
print('\nMA-RI Descriptive Summary:')
print(ma[['total_caseload', 'resolved_cases', 'pending_cases', 'deciding_productivity_rate', 'standard_clearance_rate']].describe().round(2))
```

## 7. Caseload Trends, 8. Clearance Analysis & 9. Visualizations


```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(ma['report_year'], ma['total_caseload'], marker='o', label='Total Caseload')
ax.plot(ma['report_year'], ma['resolved_cases'], marker='s', linestyle='--', label='Resolved Cases')
ax.set_title('MA-RI Caseload vs. Resolved Cases (2015-2025)')
ax.set_ylabel('Cases')
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.close('all')
```

## 10. Interpretation, 11. Limitations, 12. Policy Implications & 13. Reproducibility Links
- **Interpretation**: MA-RI caseload grew +107.30% (2015–2025) while year-end pending fell -95.57%.
- **Limitations**: Aggregate tier counts; 2015 incoming unverified (`NULL`); 3 flagged source discrepancies.
- **Policy Implications**: Pair *Rasio Produktivitas Memutus* with Standard Clearance Rate (`Resolved/Incoming`) and minutation timeliness.
